````python
#average salary across the entire company.
average_salary = df["Monthly_Salary"].mean()

average_salary
````

````python
#department with the highest average salary.
department_salary = df.groupby("Department")["Monthly_Salary"].mean()
department_salary.sort_values(ascending=False)
````

````python
#highest average salary
department_salary.idxmax()
````

````python
df["High_Performer"] = df["Performance_Score"] >= 7
df
````

````python
#Count the total number of High Performers in the company

high_performers = df[df["High_Performer"] == True]
high_performers
````

````python
# Count Total High Performers
df["High_Performer"].sum()
````

````python
#Breakdown by department:
df.groupby("Department")["High_Performer"].sum()
````

````python
#Experience vs Salary
plt.scatter(df["Years_of_Experience"], df["Monthly_Salary"])
plt.title("Experience vs Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Monthly Salary")
plt.grid(True)
plt.show()
````

````python
#Department vs Average Performance Score.

department_score = df.groupby("Department")["Performance_Score"].mean()
department_score
````

````python
department_score.plot(kind="bar")

plt.title("Department vs Average Performance Score")
plt.xlabel("Department")
plt.ylabel("Average Performance Score")

plt.xticks(rotation=45)
plt.grid(axis="y")
plt.show()
````
