````python
import pandas as pd
import matplotlib.pyplot as plt
````

````python
df = pd.read_csv("ecommerce_sales_dataset.csv")
````

````python
df
````

````python
df.info()
````

````python
#Calculate Revenue
df["Total_Sales"] = df["Price"] * df["Quantity"]
df
````

````python
#Identify Top Performers

category_sales = df.groupby("Product_Category")["Total_Sales"].sum()

category_sales.sort_values(ascending=False)
````

````python
#highest-selling category:
category_sales.idxmax()
````

````python
#average product rating.

average_rating = df["Customer_Rating"].mean()
average_rating
round(average_rating, 2)
````

````python
premium_products = df[df["Customer_Rating"] > 4]

premium_products
````

````python
#Category vs Total Sales.
category_sales.plot(kind="bar")
plt.title("Category vs Total Sales")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.show()
````

````python
#Pricing Analysis
plt.hist(df["Price"], bins=5, edgecolor="black")
plt.title("Product Price Distribution")
plt.xlabel("Product Price")
plt.ylabel("Number of Orders")
plt.grid(axis="y")
plt.show()
````
