````python
#average daily steps across all users.
average_steps = df["Daily_Steps"].mean()
average_steps
````

````python
#users who walk more than 10,000 steps per day
power_users = df[df["Daily_Steps"] > 10000]
power_users
````

````python
#Count the power users:
power_users.shape[0]
````

````python
#Step Distribution

plt.hist(df["Daily_Steps"], bins=5, edgecolor="black")
plt.title("Distribution of Daily Steps")
plt.xlabel("Daily Steps")
plt.ylabel("Number of Users")

plt.grid(axis="y")
plt.show()
````

````python
#Body Metrics Relationship

plt.scatter(df["Weight_kg"], df["Height_cm"])
plt.title("Weight vs Height")
plt.xlabel("Weight (kg)")
plt.ylabel("Height (cm)")

plt.grid(True)
plt.show()
````

````python
#Creating a new column called Active_User

df["Active_User"] = df["Daily_Steps"] >= 8000
df
````

````python
#Count how many users are Active and Inactive

active_users = df[df["Active_User"] == True]

active_users.shape[0]
````

````python
inactive_users = df[df["Active_User"] == False]

inactive_users.shape[0]
````

````python
active_percentage = (active_users.shape[0] / df.shape[0]) * 100

inactive_percentage = (inactive_users.shape[0] / df.shape[0]) * 100

active_percentage
````
